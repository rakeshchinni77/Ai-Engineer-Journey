## Machine Learning (ML)
Machine Learning is a part of AI where machines learn from data instead of being explicitly programmed.

Instead of writing rules manually, we give data — and the system finds patterns by itself.

# Real-Life Analogy

Imagine teaching a child to recognize mangoes 🍋 (okay, mango not lemon 😄).

You don’t give rules like:

If yellow + oval + sweet smell = mango

Instead, you show many mango pictures.
Over time, the child learns the pattern.

That learning from examples = Machine Learning.

# Interview Answer (1 Minute)

Structured data is organized in a fixed format like tables with rows and columns, making it easy to query and analyze.

Unstructured data has no predefined format, such as images, videos, text, and audio, and typically requires deep learning techniques.

Behavioral data captures user actions over time, like clicks, purchases, and browsing history, and is widely used for recommendations and personalization.

Graph data models entities and their relationships using nodes and edges, making it ideal for social networks, fraud detection, and recommendation systems.

# 1. Structured Data

## Definition

Structured data is organized in rows and columns with a fixed format.

Think of it like an Excel sheet or a database table.

## Example

| Student ID | Name | Age | Marks |
|------------|------|-----|-------|
| 101 | Satish | 21 | 90 |
| 102 | Ravi | 22 | 85 |

Every row has the same columns.

## Characteristics

- Fixed schema
- Easy to search
- Stored in SQL databases
- Numerical and categorical values

## Examples

- Bank transactions
- Employee records
- Student database
- Sales data

## AI/ML Use Cases

- Predict student marks
- House price prediction
- Fraud detection
- Salary prediction

## Algorithms

- Linear Regression
- Logistic Regression
- Decision Trees
- Random Forest
- XGBoost

# 2. Unstructured Data

## Definition

Unstructured data has no predefined format.

Computers cannot directly understand it.

### Examples

- Images
- Videos
- Audio
- Emails
- PDFs
- Social media posts

For example:

Image of a cat 🐱

A computer only sees pixels, not "cat."

## Characteristics

- No rows and columns
- Difficult to process directly
- Usually very large

## AI/ML Use Cases

- Face recognition
- Speech recognition
- Object detection
- Chatbots
- Medical image analysis

### Algorithms

- CNN (for images)
- Transformers (for text)
- RNN/LSTM (older sequence models)
- Vision Transformers (ViT)

# 3. Behavioral Data

## Definition

Behavioral data records what users do, not just who they are.

It focuses on actions and interactions.

## Examples

Suppose you visit YouTube.

It records:

- Which videos you watched
- How long you watched
- Which videos you skipped
- What you searched
- What you liked
- When you watched

This is behavioral data.

## Characteristics

- Time-based
- Generated continuously
- User activity logs

## Examples

- Click history
- Purchase history
- Search history
- Website navigation
- Mobile app usage

## AI/ML Use Cases

- Recommendation systems (Netflix, YouTube, Amazon)
- Customer churn prediction
- Personalized ads
- Fraud detection

## Algorithms

- Collaborative Filtering
- Deep Learning
- Reinforcement Learning
- Sequential Models

# 4. Graph Data

## Definition

Graph data represents relationships between objects.

Instead of tables, it uses:

- Nodes (entities)
- Edges (relationships)

## Example

```text
Satish ---- Friend ---- Ravi
   |
 Likes
   |
Machine Learning
```

Here:

- Satish = Node
- Ravi = Node
- Friend = Edge

## Characteristics

- Focuses on connections
- Excellent for network analysis

## Examples

- Facebook friends
- LinkedIn connections
- Road maps
- Knowledge graphs
- Supply chain networks

## AI/ML Use Cases

- Friend recommendation
- Fraud detection
- Social network analysis
- Route optimization
- Drug discovery

## Algorithms

- Graph Neural Networks (GNN)
- Graph Convolution Networks (GCN)
- Node2Vec
- PageRank

# 1. Predict

## Meaning

Prediction means estimating a future or unknown numerical value.

The output is usually a continuous number.

## Example

Suppose you have this data:

| House Size (sq ft) | Price (₹) |
|--------------------|-----------|
| 1000 | 40 Lakhs |
| 1200 | 48 Lakhs |
| 1500 | 60 Lakhs |

Now someone asks:

**What will be the price of a 1300 sq ft house?**

The model predicts:

➡ ₹52 Lakhs

This is prediction.

## AI Algorithms

- Linear Regression
- Random Forest Regressor
- XGBoost Regressor

## Other Examples

- Predict stock price
- Predict rainfall
- Predict salary
- Predict student marks

# 2. Classify

## Meaning

Classification means assigning data to a predefined category or label.

The output is discrete, not a number.

## Example

**Email:**

> "Congratulations! You won ₹10 Lakhs."

**Question:**

Spam or Not Spam?

**Output:**

➡ Spam

Another example:

**Patient symptoms**

**Output:**

➡ Diabetic / Not Diabetic

## AI Algorithms

- Logistic Regression
- Decision Tree
- SVM
- Random Forest
- Neural Networks

## Other Examples

- Cat vs Dog
- Fraud vs Genuine Transaction
- Positive vs Negative Review

3. Cluster
Meaning

Clustering means grouping similar data together without knowing the labels beforehand.

This is Unsupervised Learning.

Example:

You have 100 customers.

You don't know anything about them.

The algorithm automatically groups them:

Group A → Students

Group B → Working Professionals

Group C → Senior Citizens

Nobody told the model these labels—it discovered the groups itself.

AI Algorithms
K-Means
DBSCAN
Hierarchical Clustering
Real-world examples
Customer segmentation
Friend suggestions
Product grouping

# Interview Answer

The goal of an AI/ML system determines the type of problem and algorithm to use.

- **Prediction** estimates a continuous numerical value, such as house prices.
- **Classification** assigns data to predefined categories, such as spam or not spam.
- **Clustering** groups similar data without predefined labels.
- **Detection** identifies the presence or location of objects, events, or anomalies, such as helmet or fraud detection.
- **Generation** creates new content like text, images, audio, or code.
