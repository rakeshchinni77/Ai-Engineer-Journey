## Linear Regression

### 1. What is it?
Linear Regression is a supervised machine learning algorithm used to predict a continuous numerical value by finding the best-fit straight line that represents the relationship between input features (independent variables) and the target variable (dependent variable).

It assumes that there is a linear relationship between inputs and output.

Mathematical Form

For one feature:

y=mx+c

Machine Learning form:

y=w1​x1​+w2​x2​+...+wn​xn​+b

x = Input feature
w = Weight/Coefficient
b = Bias/Intercept
y = Predicted output
